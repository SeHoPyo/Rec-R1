```python
import pandas as pd

# 파일 경로 설정
train_parquet = "/Users/shingeunbang/RLproj/Rec-R1_3/Rec-R1/data/amazon_c4/sports_parquet/train.parquet"
val_parquet = "/Users/shingeunbang/RLproj/Rec-R1_3/Rec-R1/data/amazon_c4/sports_parquet/val.parquet"
test_parquet = "/Users/shingeunbang/RLproj/Rec-R1_3/Rec-R1/data/amazon_c4/sports_parquet/test.parquet"

train_csv = "/Users/shingeunbang/RLproj/Rec-R1_3/Rec-R1/data/amazon_c4/sports_parquet/csv_file/train.csv"
val_csv = "/Users/shingeunbang/RLproj/Rec-R1_3/Rec-R1/data/amazon_c4/sports_parquet/csv_file/val.csv"
test_csv = "/Users/shingeunbang/RLproj/Rec-R1_3/Rec-R1/data/amazon_c4/sports_parquet/csv_file/test.csv"

# Parquet 파일을 읽어서 CSV로 저장
df_train = pd.read_parquet(train_parquet)
df_val = pd.read_parquet(val_parquet)
df_test = pd.read_parquet(test_parquet)

df_train.to_csv(train_csv, index=False)
df_val.to_csv(val_csv, index=False)
df_test.to_csv(test_csv, index=False)

print("train, val, test 모두 csv 파일로 저장 완료!")
```
